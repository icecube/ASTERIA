# Signal per DOM

Load a simulation and plot the signal per DOM.

```python
from asteria.interactions import Interactions
from asteria.simulation import Simulation

from snewpy.neutrino import Flavor

from importlib.resources import files

import astropy.units as u

import numpy as np
import matplotlib as mpl
import matplotlib.pyplot as plt
```

## Set up styles for plotting

```python
mpl.rc('font', size=14)
```

## Load Configuration and Run Simulation

This will load the simulation configuration from a file.

In this case, we'll use a default INI file distributed with `ASTERIA`. Edit or update the configuration file as desired.

```python
conf = files('asteria.etc').joinpath('example.ini')
sim = Simulation(configfile=conf)
sim.run()
```

## Compute Signal per DOM

```python
effvol = 0.1654 * u.m**3 / u.MeV

E_per_V = sim.E_per_V
signal_per_DOM = { flavor: effvol*E_per_V[flavor] for flavor in Flavor }
```

## Plot Signal Increase Per DOM
This will plot the signal increase in a single DOM caused by each flavor. The curve shows the DOM response from 0.05 seconds before the core bounce to 0.65 seconds after the core bounce. Note that this is computed with the assumption that the effective volume of the DOM is 0.1654 m$^3$ MeV$^{-1}$

```python
fig, ax = plt.subplots(1, figsize = (7,5), tight_layout=True)

for flavor in Flavor:
    width = 1 if flavor.is_neutrino else 3
    style = '-' if flavor.is_neutrino else ':'
    color = 'tab:orange' if flavor.is_electron else ('tab:blue' if flavor.is_muon else 'tab:red')
    ax.step(sim.t, signal_per_DOM[flavor], lw=width, ls=style, color=color, label=flavor.to_tex())
    
ax.set_ylabel( 'Signal per DOM', horizontalalignment='right', y = 1)
ax.set_xlabel(r't-t$_{bounce}$ [s]', horizontalalignment='right', x=1.0)
ax.set(xlim=(-0.05, 1))
ax.legend(fontsize=12, ncol=3)
ax.grid(ls=':')
```
